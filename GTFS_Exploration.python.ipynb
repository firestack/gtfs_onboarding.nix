```python
%%sql
SELECT
	CAST(stop.stop_lon AS REAL),
	CAST(stop.stop_lat AS REAL)
FROM
	stop_times
		AS stop_time
JOIN
	stops
		AS stop
	ON
		stop_time.stop_id
			= stop.stop_id
WHERE
	stop_time.trip_id
		= '60487527'
ORDER BY
	stop_sequence;
```
